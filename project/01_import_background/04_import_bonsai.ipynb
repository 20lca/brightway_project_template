# BONSAI import

This notebook imports BONSAI into the configured Brightway project.

Project names, database names and file paths are defined in `functions/utils.py`.
The BONSAI-specific biosphere mapping is stored in `functions/bonsai_import.py`.


### 1. Import required packages


```python
import bw2data as bd
import bw2io as bi

from bw2io.importers.bonsai import BonsaiImporter

from functions.bonsai_import import define_b3_map
from functions.utils import Config
```

### 2. Select Brightway project


```python
bd.projects.set_current(Config.PROJECT_NAME)

print(f"Current Brightway project: {bd.projects.current}")
```

### 3. Check BONSAI input folder

The BONSAI folder should contain the files required by `BonsaiImporter`.


```python
assert Config.BONSAI_PATH.is_dir(), (
    f"BONSAI folder not found at: {Config.BONSAI_PATH}"
)

print(f"BONSAI folder: {Config.BONSAI_PATH}")
```

### 4. Create/check biosphere database

BONSAI is linked to the configured Brightway biosphere database.


```python
if Config.BONSAI_BIOSPHERE_NAME not in bd.databases:
    bi.create_default_biosphere3()

print(f"Using biosphere database: {Config.BONSAI_BIOSPHERE_NAME}")
```

### 5. Remove an existing BONSAI import

Delete the BONSAI databases if they already exist so the importer can be run again cleanly.


```python
if Config.BONSAI_DATABASE_NAME in bd.databases:
    bd.Database(Config.BONSAI_DATABASE_NAME).delete(warn=False)

if Config.BONSAI_BIOSPHERE_DATABASE_NAME in bd.databases:
    bd.Database(Config.BONSAI_BIOSPHERE_DATABASE_NAME).delete(warn=False)
```

### 6. Create the BONSAI biosphere mapping

The BONSAI importer needs a small mapping between BONSAI flow names and flows in the Brightway biosphere database.


```python
mapping_b3 = define_b3_map()
mapping_b3
```

### 7. Create the BONSAI importer


```python
bonsai = BonsaiImporter(
    Config.BONSAI_PATH,
    Config.BONSAI_DATABASE_NAME,
    b3mapping=mapping_b3,
)
```

### 8. Apply import strategies


```python
bonsai.apply_strategies()
```

### 9. Write the BONSAI database

This writes the processed BONSAI database to the current Brightway project.


```python
bonsai.write_database()
```
