### 3. Option A = install ecoinvent with username and password (preferred method)

Note that there are different ways of importing ecoinvent to your brightway project. If you have username and password at hand you can install the database just using this access codes. The first suggestion is to use a structured importer available from the brightway cloud.

```python
bi.import_ecoinvent_release(
    version = Config.ECOINVENT_VERSION,
    system_model = Config.ECOINVENT_SYSTEM_MODEL,
    username = Config.ECOINVENT_USERNAME,
    password = Config.ECOINVENT_PASSWORD
    )
```

### 4. Option B = install ecoinvent with Ecospold files (legacy method)

```python
# Check whether the standard Brightway biosphere database is available
if "biosphere3" not in bd.databases:
    bi.create_default_biosphere3()
```

```python
# Create importer
importer = SingleOutputEcospold2Importer(
    Config.ECOINVENT_ECOSPOLD_PATH,
    Config.ECOINVENT_CLCA_NAME,
)
```

```python
# Apply import strategies
importer.apply_strategies()
```

```python
# Inspect unlinked exchanges
pd.DataFrame(importer.unlinked)
```

```python
# If no unlinked exchanges are found, write the database to the Brightway project 
importer.write_database()
```
