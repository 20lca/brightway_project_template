## 2. Option A: install ecoinvent with username and password

Note that there are different ways of importing ecoinvent to your brightway project. If you have username and password at hand you can install the database just using this access codes. The first suggestion is to use a structured importer available from brightway.

```python
bi.import_ecoinvent_release(
    version = Config.ECOINVENT_VERSION,
    system_model = Config.ECOINVENT_SYSTEM_MODEL,
    username = Config.ECOINVENT_USERNAME,
    password = Config.ECOINVENT_PASSWORD
    )
```

## 3. Option B: install ecoinvent with Ecospold files

```python
# Install the relevant biosphere database
biosphere_importer = Ecospold2BiosphereImporter(
    version = Config.ECOINVENT_VERSION,
    filepath = Config.ECOINVENT_BIOSPHERE_PATH,
    name = Config.ECOINVENT_BIOSPHERE_NAME
)
```

```python
biosphere_importer.apply_strategies()
```

```python
pd.DataFrame(biosphere_importer.unlinked)
```

```python
biosphere_importer.write_database()
```

```python
# Create importer
importer = bi.SingleOutputEcospold2Importer(
    Config.ECOINVENT_ECOSPOLD_PATH,
    Config.ECOINVENT_CLCA_NAME,
)
```

```python
# Inspect unlinked exchanges
pd.DataFrame(importer.unlinked)
```
